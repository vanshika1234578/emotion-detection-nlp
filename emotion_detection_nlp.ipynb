WE HAVE TO CONVERT EACH EMOTION INTO NUMERIC TERMS
WE MAKE A DICT .. TO SAVE EMOTIONS

```python
unique_emo=df['emotion'].unique()
emotion_num={}
for i in range(len(unique_emo)):
    emotion_num[unique_emo[i]]=i
emotion_num
df['emotion']=df['emotion'].map(emotion_num)
df
```

```python
df['text']=df['text'].apply(lambda x:x.lower())
df
```

#REMOVE PUNCTUATIONS

```python
import string
def rem_pun(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']= df['text'].apply(rem_pun)
df
```

REMOVE NUMBERS

```python
def rev_num(txt):
  new=""
  for i in txt:
    if not i.isdigit():
      new=new+i
  return new
```

```python
df['text']=df['text'].apply(rev_num)
df
```

HOW TO REMOVE EMOJIS
EMOJIS HAVE UNICODE ONLY

```python
def rem_em(txt):
  new=""
  for i in txt:
    if i.isascii():
      new=new+i
  return new
```

```python
df['text']=df['text'].apply(rem_em)
df
```

REMOVE STOP WORDS

```python
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('stopwords')
nltk.download('punkt')# does tokenisation
```

```python
stop_words=set(stopwords.words('english'))
stop_words
```

```python
len(stop_words)
```

```python
df.loc[1]['text']
```

CONVERT THE TEXT TO ARRAY AND COMPARE WITH STOP WORDS : THIS IS TOKENISATION

```python
def rem_stop(txt):
  words=word_tokenize(txt)
  new=[]
  for i in words:
    if i not in stop_words:
      new.append(i)
  return " ".join(new)
```

```python
import nltk
nltk.download('punkt_tab')

df['text']=df['text'].apply(rem_stop)
df
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df['text'],df['emotion'],test_size=0.2,random_state=42)
# use bow and TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer
vectoriser=TfidfVectorizer()
BOW_vectoriser=CountVectorizer()
x_train_tf=vectoriser.fit_transform(x_train)
x_test_tf=vectoriser.transform(x_test)
x_train_bow=BOW_vectoriser.fit_transform(x_train)
x_test_bow=BOW_vectoriser.transform(x_test)

x_train_tf
```

model we will use is naive bayes , logisitic reg.

```python
from sklearn.naive_bayes import MultinomialNB #gassian is not used , we use multinomial
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
nb_model_tf=MultinomialNB()
nb_model_tf.fit(x_train_tf,y_train)
y_pred_tf=nb_model_tf.predict(x_test_tf)
print("tf accu:",accuracy_score(y_test,y_pred_tf))

nb_model_bow=MultinomialNB()
nb_model_bow.fit(x_train_bow,y_train)
y_pred_bow=nb_model_bow.predict(x_test_bow)
print("bow accu:",accuracy_score(y_test,y_pred_bow))
```

```python
log_model_tf=LogisticRegression(max_iter=1000)
log_model_tf.fit(x_train_tf,y_train)
y_pred_tf=log_model_tf.predict(x_test_tf)
print("tf accu:",accuracy_score(y_test,y_pred_tf))
```
